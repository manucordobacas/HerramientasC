# primer ejercicio

```python

#entrada

nombre_cli=input("por favor escriba el nombre del cliente")
precio1=float(input("por favor escriba el precio del producto 1"))
precio2=float(input("por favor escriba el precio del producto 2"))
precio3=float(input("por favor escriba el precio del producto 3"))
cantidad3=int(input("por favor escriba la cantidad del producto 3"))
descuento=float(input("por favor escriba el descuento"))

#proceso

subtotal1y2=precio1+precio2
valortotal3=precio3*cantidad3
subtotal=subtotal1y2+valortotal3
total = subtotal-(subtotal*(descuento/100))
promedio=(valortotal3+subtotal1y2)/3
valorvdd1=bool(subtotal>=500000)
valorvvdd2=bool(descuento>10)
valorvdd3=bool(total<subtotal)

#salida
print("Hola", nombre_cli)
print("el subtotal supera los 500.000?", valorvdd1)
print("El descuento es mayor a 10%?", valorvvdd2)
print("El total es menor que el subtotal?", valorvdd3)
print("la suma de los precios del producto 1 y 2 es:", subtotal1y2)
print("el valor total del producto 3 es:", valortotal3)
print("El subtotal (teniendo en cuenta los productos 1,2 y 3) es:", subtotal)
print("el total(o sea incluyendo el descuento) es:", total)
print("el promedio de los precios de los 3 productos es:", round(promedio,2)) 



```

```python
# entrada
nombre_estudiante=input("por favor escriba el nombre del estudiante")
nota1=float(input("por favor escriba la nota del primer parcial"))
nota2=float(input("por favor escriba la nota del segundo parcial"))
nota3=float(input("por favor escriba la nota del tercer parcial"))

#proceso
nota1vdd=bool(nota1>=0 and nota1<=5)
nota2vdd=bool(nota2>=0 and nota2<=5)
nota3vdd=bool(nota3>=0 and nota3<=5)
promedio_notas=(nota1+nota2+nota3)/3
notamax=max(nota1,nota2,nota3)
notamin=min(nota1,nota2,nota3)
difnota=notamax-notamin
promediovdd=bool(promedio_notas>=3.5)
nota14vdd=bool(nota1>=4)
nota24vdd=bool(nota2>=4)
nota34vdd=bool(nota3>=4)
resultado4vdd=bool(nota14vdd+nota24vdd+nota34vdd>=2)
nota13vdd=bool(nota1>=3.0)
nota23vdd=bool(nota2>=3.0)
nota33vdd=bool(nota3>=3.0)
resultado3vdd=bool(nota13vdd+nota23vdd+nota33vdd==3)
nota15vdd=bool(nota1==5.0)
nota25vdd=bool(nota2==5.0)
nota35vdd=bool(nota3==5.0)
resultado5vdd=bool(nota15vdd or nota25vdd or nota35vdd)

#salida
print("Hola",nombre_estudiante,"sus notas son:",nota1,nota2,nota3,"y su promedio es:", promedio_notas) 
print("¿su nota 1 está en el rango establecido?", nota1vdd)
print("¿su nota 2 está en el rango establecido?", nota2vdd)
print("¿su nota 3 está en el rango establecido?", nota3vdd)
print("su nota máxima es:", notamax)
print("su nota mínima es:", notamin)
print("La diferencia entre la nota máxima y mínima es:", difnota)
print("¿al menos 2 de sus notas son mayores o iguales a 4?", resultado4vdd)
print("¿todas sus notas son mayores o iguales a 3?", resultado3vdd)
print("¿al menos una de sus notas es igual a 5?", resultado5vdd)

```

```python
#entrada
segundos=int(input("ingrese los segundos que desea convertira a horas, minutos, segundos"))

#proceso
horas=segundos//3600
minutos=(segundos%3600)//60
segundos_rest=segundos%60
reconstruccion=horas*3600+minutos*60+segundos_rest
diferencia=segundos-reconstruccion
cantidadvdd=bool(segundos>3600)
cantidadvdd2=bool(segundos>86400)
vdd3=bool(segundos%60==0)

#salida
print("la cantidad de horas es:", horas)
print("la cantidad de minutos es:", minutos)
print("la cantidad de segundos es:", segundos_rest) 
print("la diferencia entre los segundos ingresados y la reconstrucción es:", diferencia)
print("¿la cantidad de tiempo es mayor a 1 hora(3600s)?", cantidadvdd)
print("¿la cantidad de tiempo es mayor a un dia(86400s)?", cantidadvdd2)
print("¿la cantidad es exactamente divisible entre 60?", vdd3)


```
